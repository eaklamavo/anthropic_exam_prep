## Domain Overview

This domain covers using Claude to analyze business requirements, support research and
planning, help design and iterate on solutions, integrate Claude into existing workflows, and
communicate its value and limitations to stakeholders.

**Why it matters**: Associates are expected to go beyond single prompts and actually improve how
their team works — spotting where Claude fits into a process and explaining that clearly to others.

**Key concepts to know**: requirements analysis, workflow redesign vs. augmentation, and
stakeholder communication about AI capabilities and limits.

```python
# Setup: a small helper so every cell in this notebook runs end-to-end,
# whether or not you have an ANTHROPIC_API_KEY configured.
import os

def call_claude(prompt, system=None, model="claude-sonnet-4-5", max_tokens=600):
    """Calls the real Claude API if ANTHROPIC_API_KEY is set; otherwise returns a
    simulated response so this notebook is runnable without API access."""
    try:
        import anthropic
        api_key = os.environ.get("ANTHROPIC_API_KEY")
        if not api_key:
            raise RuntimeError("no ANTHROPIC_API_KEY set")
        client = anthropic.Anthropic(api_key=api_key)
        kwargs = {"model": model, "max_tokens": max_tokens, "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        return (f"[SIMULATED RESPONSE - set ANTHROPIC_API_KEY and `pip install anthropic` "
                f"to call the real API]\nPrompt received ({len(prompt)} chars): {prompt[:150]}...")

print("call_claude() ready. Set ANTHROPIC_API_KEY as an environment variable to hit the real API.")
```

## Key Concepts

### Apply Claude to analyze requirements and use cases

Before building anything, use Claude to clarify what a workflow actually needs: current pain
points, stakeholders, inputs/outputs, and success criteria. This turns a vague ask ("make this
faster") into a concrete requirements list.

```python
vague_ask = "Make our client onboarding process faster."
requirements = call_claude(f"""Given this request: "{vague_ask}"
List the requirements-analysis questions you would need answered before proposing a solution.""")
print(requirements)
```

### Leverage Claude for research, planning, and process optimization

Claude can synthesize research on best practices, help sequence a project plan, and
identify bottlenecks in an existing process description.

```python
current_process = "1) Sales hands off deal. 2) Ops manually creates account. 3) Ops emails welcome kit. 4) Client waits 5 days for setup call."
optimization = call_claude(f"Identify the top 2 bottlenecks in this process and suggest a fix for each: {current_process}")
print(optimization)
```

### Use Claude to support solution design, development, and iteration

Claude can help draft a solution outline, then iterate on it as constraints emerge (budget,
tooling, team size) — similar to Domain 1's iteration skill but applied to solution architecture rather
than a single document.

```python
solution_v1 = call_claude("Propose a solution to reduce onboarding time from 5 days to 1 day, given the process above.")
print(solution_v1)

constraint_feedback = "We have no budget for new software this quarter. Revise assuming manual/process-only changes."
solution_v2 = call_claude(f"{solution_v1}\n\n{constraint_feedback}")
print(solution_v2)
```

### Integrate Claude into existing workflows to augment or redesign them

Augmenting a workflow keeps the existing structure and adds Claude at one step (e.g., Claude
drafts the welcome email). Redesigning rethinks the workflow itself (e.g., eliminating a manual step
entirely because Claude can trigger it).

```python
integration_options = {
    "augment": "Keep the 4-step process, but Claude auto-drafts the welcome email in step 3.",
    "redesign": "Claude reviews the deal handoff and auto-populates the account, skipping steps 2-3 manually.",
}
for approach, description in integration_options.items():
    print(f"{approach}: {description}")
```

### Communicate Claude's value and limitations to stakeholders

Stakeholders need both sides: what Claude reliably improves (speed, consistency, first-draft
quality) and where it still needs oversight (accuracy on edge cases, judgment calls, anything
high-stakes).

```python
stakeholder_summary = call_claude("""
Write a 3-sentence summary for a department head explaining: what Claude will speed up in our
onboarding process, and one limitation/oversight need they should be aware of.
""")
print(stakeholder_summary)
```

## Answers


**A1.** Because a vague request without clear requirements risks solving the wrong problem or missing success criteria.

**A2.** Augmenting keeps the existing process structure and inserts Claude at one step; redesigning rethinks the workflow itself.

**A3.** Iterate the existing proposal with the new constraint as explicit feedback, rather than starting over from scratch.

**A4.** Both the concrete value/benefit AND the specific limitations or oversight needs, not just one side.

**A5.** False — sometimes a process change (e.g., eliminating a manual step) is more effective and cheaper than new software.
