## Domain Overview

This domain covers diagnosing why a prompt or workflow isn't producing good output,
adjusting your approach based on feedback and results, and optimizing workflows over time for
efficiency and effectiveness.

**Why it matters**: even well-designed prompts and workflows degrade or underperform sometimes;
Associates need a systematic way to diagnose and fix that rather than guessing.

**Key concepts to know**: root-cause diagnosis of poor outputs, feedback-driven adjustment, and
ongoing workflow optimization.

```python
# Setup: a small helper so every cell in this notebook runs end-to-end,
# whether or not you have an ANTHROPIC_API_KEY configured.
import os

def call_claude(prompt, system=None, model="claude-sonnet-4-5", max_tokens=600):
    """Calls the real Claude API if ANTHROPIC_API_KEY is set; otherwise returns a
    simulated response so this notebook is runnable without API access."""
    try:
        import anthropic
        api_key = os.environ.get("ANTHROPIC_API_KEY")
        if not api_key:
            raise RuntimeError("no ANTHROPIC_API_KEY set")
        client = anthropic.Anthropic(api_key=api_key)
        kwargs = {"model": model, "max_tokens": max_tokens, "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        return (f"[SIMULATED RESPONSE - set ANTHROPIC_API_KEY and `pip install anthropic` "
                f"to call the real API]\nPrompt received ({len(prompt)} chars): {prompt[:150]}...")

print("call_claude() ready. Set ANTHROPIC_API_KEY as an environment variable to hit the real API.")
```

## Key Concepts

### Identify, diagnose, and resolve issues with underperforming prompts or poor outputs

Common root causes of poor output: missing context, ambiguous instructions, wrong format
request, or a task that's too complex for a single prompt (needs decomposition). Diagnosis means
identifying which of these applies before attempting a fix.

```python
poor_output = call_claude("fix this")  # deliberately vague/under-specified
print(poor_output)

diagnosis_checklist = {
    "missing context": True,
    "ambiguous instruction": True,
    "wrong format requested": False,
    "task too complex for one prompt": False,
}
print(diagnosis_checklist)
```

### Adjust approach based on feedback and results

Once a root cause is identified, the fix should target that cause specifically: add context if
missing, clarify instructions if ambiguous, specify format if wrong, or decompose if too complex.

```python
fixed_prompt = """
Context: This is a broken CSS layout where the navbar overlaps the hero image on mobile.
Task: Suggest 2 specific CSS fixes.
Format: Numbered list, each with a one-line explanation.
"""
print(call_claude(fixed_prompt))
```

### Optimize workflows for efficiency and effectiveness

Beyond fixing single prompts, optimization looks at the whole workflow: are there
repeated manual steps that could be templated? Is the same context being re-typed every time
when it could live in a Project's instructions instead?

```python
workflow_before = {"steps": 6, "manual_context_retyping": True, "template_used": False}
workflow_after = {"steps": 4, "manual_context_retyping": False, "template_used": True}
print("Efficiency gain: fewer steps and no repeated manual context entry:", workflow_after)
```
