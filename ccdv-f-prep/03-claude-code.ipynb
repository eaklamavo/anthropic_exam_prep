```python
# Setup: a small helper so every cell in this notebook runs end-to-end,
# whether or not you have an ANTHROPIC_API_KEY configured.
import os

def call_claude(prompt, system=None, model="claude-sonnet-4-5", max_tokens=600):
    """Calls the real Claude API if ANTHROPIC_API_KEY is set; otherwise returns a
    simulated response so this notebook is runnable without API access."""
    try:
        import anthropic
        api_key = os.environ.get("ANTHROPIC_API_KEY")
        if not api_key:
            raise RuntimeError("no ANTHROPIC_API_KEY set")
        client = anthropic.Anthropic(api_key=api_key)
        kwargs = {"model": model, "max_tokens": max_tokens, "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        return (f"[SIMULATED RESPONSE - set ANTHROPIC_API_KEY and `pip install anthropic` "
                f"to call the real API]\nPrompt received ({len(prompt)} chars): {prompt[:150]}...")

print("call_claude() ready. Set ANTHROPIC_API_KEY as an environment variable to hit the real API.")
```

## Key Concepts

### Claude Code core components: Rules, Skills, Commands, Agents, Agent Memory

Rules (`.claude/rules/`) apply conditional conventions by file path. Skills
(`.claude/skills/`) are on-demand, packaged capabilities. Commands (`.claude/commands/`) are custom
slash commands. Agents are subagent definitions for delegated work. Agent Memory persists facts
across a session or across sessions depending on configuration.

```python
core_components = {
    "Rules": ".claude/rules/ - conditional conventions by file path (YAML frontmatter globs)",
    "Skills": ".claude/skills/ - on-demand packaged capabilities (SKILL.md)",
    "Commands": ".claude/commands/ - custom slash commands",
    "Agents": "subagent definitions for delegated, scoped work",
    "Agent Memory": "persists key facts across a session (or across sessions if configured)",
}
for k, v in core_components.items():
    print(f"{k:12s}: {v}")
```

### Session management, headless mode, streaming mode, auto-mode

Headless mode (`-p`/`--print`) runs non-interactively for scripts/CI. Streaming mode
outputs incrementally for responsive terminals/UIs. Auto-mode allows Claude Code to proceed through
routine actions without per-step confirmation, appropriate for well-scoped, low-risk tasks.

```python
def choose_claude_code_mode(interactive: bool, needs_ci_automation: bool) -> str:
    if needs_ci_automation:
        return "-p / --print (headless/non-interactive mode)"
    if not interactive:
        return "streaming mode for a responsive but non-blocking terminal experience"
    return "standard interactive session"

print(choose_claude_code_mode(interactive=False, needs_ci_automation=True))
```

### CLAUDE.md hierarchy and repository initialization

CLAUDE.md exists at user (`~/.claude/CLAUDE.md`), project (`.claude/CLAUDE.md` or root),
and directory levels. Repository initialization sets up the baseline `.claude/` structure (CLAUDE.md,
rules, commands, skills folders) for a new project.

```python
def init_claude_code_repo_structure() -> list[str]:
    return [".claude/CLAUDE.md", ".claude/rules/", ".claude/commands/", ".claude/skills/", ".mcp.json"]

print(init_claude_code_repo_structure())
```

### settings.json configuration

`settings.json` configures Claude Code behavior at the user or project level: permitted
tools, auto-mode thresholds, and other operational defaults, distinct from CLAUDE.md's role of
providing project *context* rather than tool *permissions*.

```python
settings_json_example = {
    "permissions": {"allow": ["Read", "Grep", "Glob"], "ask": ["Bash", "Write", "Edit"]},
    "autoApprove": False,
}
import json
print(json.dumps(settings_json_example, indent=2))
```
