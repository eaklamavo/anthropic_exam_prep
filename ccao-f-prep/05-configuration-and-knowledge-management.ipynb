## Domain Overview

This domain covers setting up Claude Projects properly: writing effective system-level
instructions, configuring knowledge sources and connectors (Google Drive, Gmail, etc.), and keeping
that configuration current as needs change.

**Why it matters**: a poorly configured Project produces inconsistent, generic output no matter how
good individual prompts are. Configuration is a one-time investment that pays off across every
future conversation in that Project.

**Key concepts to know**: Project instructions, knowledge source management, connector setup,
and configuration maintenance.

```python
# Setup: a small helper so every cell in this notebook runs end-to-end,
# whether or not you have an ANTHROPIC_API_KEY configured.
import os

def call_claude(prompt, system=None, model="claude-sonnet-4-5", max_tokens=600):
    """Calls the real Claude API if ANTHROPIC_API_KEY is set; otherwise returns a
    simulated response so this notebook is runnable without API access."""
    try:
        import anthropic
        api_key = os.environ.get("ANTHROPIC_API_KEY")
        if not api_key:
            raise RuntimeError("no ANTHROPIC_API_KEY set")
        client = anthropic.Anthropic(api_key=api_key)
        kwargs = {"model": model, "max_tokens": max_tokens, "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        return (f"[SIMULATED RESPONSE - set ANTHROPIC_API_KEY and `pip install anthropic` "
                f"to call the real API]\nPrompt received ({len(prompt)} chars): {prompt[:150]}...")

print("call_claude() ready. Set ANTHROPIC_API_KEY as an environment variable to hit the real API.")
```

## Key Concepts

### Configure Claude Projects with instructions and knowledge sources

Project instructions set persistent behavior (tone, format, constraints) applied to every
conversation in the Project. Knowledge sources are the reference documents Claude should draw
from — both should be scoped tightly to the Project's actual purpose.

```python
project_instructions = """
You are a support-content assistant for Acme SaaS.
Always cite the specific help-article title you're drawing from.
Never suggest workarounds involving disabling security features.
Keep answers under 150 words unless asked for more detail.
"""
print(call_claude("How do I export my data?", system=project_instructions))
```

### Manage uploaded knowledge and connectors

Connectors (e.g., Google Drive, Gmail) let Claude access live, current data instead of stale
uploaded copies. Choosing between an upload and a connector depends on how often the source
data changes.

```python
sources = {
    "quarterly policy PDF (rarely changes)": "upload",
    "live shared drive of active project docs": "connector",
    "one-time competitor analysis spreadsheet": "upload",
    "team inbox for support escalations": "connector",
}
for source, method in sources.items():
    print(f"{source:45s} -> {method}")
```

### Create effective system-level instructions

Effective instructions are specific and behavioral (what to always/never do), not vague
aspirations ("be helpful"). They should state format, tone, constraints, and escalation rules.

```python
weak_instructions = "Be helpful and professional."
strong_instructions = """
Always answer in under 120 words.
If a question involves refunds over $500, respond only with: "This requires manual review - escalate to a manager."
Use a friendly but concise tone; no marketing language.
"""
print(call_claude("Can I get a $700 refund?", system=strong_instructions))
```

### Inform, maintain, and update Claude configurations, knowledge sources, and instructions

Configurations drift out of date as policies, products, and teams change. Periodic review of
instructions and knowledge sources (e.g., quarterly) keeps output accurate and prevents Claude from
citing outdated information.

```python
last_updated = "2025-01-15"
current_date = "2025-09-07"

def needs_review(last_updated: str, current_date: str, review_cycle_days: int = 90) -> bool:
    from datetime import date
    d1 = date.fromisoformat(last_updated)
    d2 = date.fromisoformat(current_date)
    return (d2 - d1).days > review_cycle_days

print(needs_review(last_updated, current_date))
```
