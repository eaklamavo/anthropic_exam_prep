```python
# Setup: a small helper so every cell in this notebook runs end-to-end,
# whether or not you have an ANTHROPIC_API_KEY configured.
import os

def call_claude(prompt, system=None, model="claude-sonnet-4-5", max_tokens=600):
    """Calls the real Claude API if ANTHROPIC_API_KEY is set; otherwise returns a
    simulated response so this notebook is runnable without API access."""
    try:
        import anthropic
        api_key = os.environ.get("ANTHROPIC_API_KEY")
        if not api_key:
            raise RuntimeError("no ANTHROPIC_API_KEY set")
        client = anthropic.Anthropic(api_key=api_key)
        kwargs = {"model": model, "max_tokens": max_tokens, "messages": [{"role": "user", "content": prompt}]}
        if system:
            kwargs["system"] = system
        resp = client.messages.create(**kwargs)
        return resp.content[0].text
    except Exception as e:
        return (f"[SIMULATED RESPONSE - set ANTHROPIC_API_KEY and `pip install anthropic` "
                f"to call the real API]\nPrompt received ({len(prompt)} chars): {prompt[:150]}...")

print("call_claude() ready. Set ANTHROPIC_API_KEY as an environment variable to hit the real API.")
```

## Key Concepts

### Conduct structured discovery and requirement gathering

Structured discovery uses a repeatable framework (current-state process, pain points,
constraints, success criteria, stakeholders) rather than ad-hoc conversation, ensuring nothing
material is missed before design begins.

```python
discovery_framework = ["current-state process", "pain points", "constraints (budget/time/tech)",
                       "success criteria", "key stakeholders", "existing systems to integrate with"]
def discovery_checklist_complete(answers: dict) -> bool:
    return all(item in answers for item in discovery_framework)

sample_answers = {k: f"answer for {k}" for k in discovery_framework}
print(discovery_checklist_complete(sample_answers))
```

### Communicate architectural decisions and trade-offs

Effective communication states the decision, the alternatives considered, why this one was
chosen, and what tradeoff was accepted (e.g., "we chose Sonnet over Opus, accepting slightly lower
ceiling accuracy, to meet the 1-second latency SLA at 1/3 the cost").

```python
def decision_summary(decision: str, alternatives: list, chosen_reason: str, accepted_tradeoff: str) -> str:
    return (f"Decision: {decision}\nAlternatives considered: {', '.join(alternatives)}\n"
            f"Why chosen: {chosen_reason}\nTradeoff accepted: {accepted_tradeoff}")

print(decision_summary(
    "Use Sonnet, not Opus, for claims triage",
    ["Opus", "Sonnet", "Haiku"],
    "Meets the 1s latency SLA at 1/3 the cost of Opus with acceptable accuracy",
    "Slightly lower ceiling accuracy on the hardest 5% of ambiguous cases",
))
```

### Manage stakeholder feedback loops and expectation alignment (including SLAs)

Feedback loops should be scheduled and structured (e.g., weekly demo + written summary),
not purely reactive. SLA alignment means stakeholders agree in writing on measurable targets
(latency, accuracy, uptime) before launch, preventing later disputes about what "good" means.

```python
sla_agreement = {"p95_latency_ms": 2000, "accuracy_target": 0.90, "uptime_target": 0.995}
def meets_sla(observed: dict, sla: dict) -> dict:
    return {k: observed[k] <= sla[k] if "latency" in k else observed[k] >= sla[k] for k in sla}

print(meets_sla({"p95_latency_ms": 1800, "accuracy_target": 0.93, "uptime_target": 0.997}, sla_agreement))
```

### Document architectures and provide implementation guidance

Architecture documentation should be usable by someone who wasn't in the room: it needs a
diagram or clear component list, the key decisions with rationale, known limitations, and concrete
implementation guidance (not just a high-level diagram with no actionable detail).

```python
architecture_doc_sections = [
    "system overview / component diagram",
    "key decisions and rationale",
    "known limitations and risks",
    "implementation guidance for engineers",
    "monitoring and escalation contacts",
]
print(architecture_doc_sections)
```

### Support lifecycle phases (discovery, design, handoff, monitoring, iteration)

Each lifecycle phase has a distinct owner and exit criteria: discovery (requirements
signed off), design (architecture approved), handoff (implementation team briefed, docs delivered),
monitoring (SLAs tracked in production), iteration (feedback drives the next design cycle).

```python
lifecycle_phases = {
    "discovery": "exit criteria: requirements document signed off by stakeholders",
    "design": "exit criteria: architecture reviewed and approved",
    "handoff": "exit criteria: implementation team briefed, docs delivered",
    "monitoring": "exit criteria: SLA dashboard live and reviewed weekly",
    "iteration": "exit criteria: feedback triaged into next design cycle backlog",
}
for phase, criteria in lifecycle_phases.items():
    print(f"{phase:12s}: {criteria}")
```
